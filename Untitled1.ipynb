```python
# ==========================================
# Python 資料型別實作作業：聚餐費用試算器
# ==========================================

def calculate_dinner_costs(event_name_raw, meal_amount_raw, people_raw, service_rate, discount, budget_per_person):
    # 1. 整理聚餐名稱：去除頭尾空白並轉成小寫
    event_name = event_name_raw.strip().lower()

    # 2. 整理餐費與人數並轉換型別
    # 移除 "NT$"、逗號與空白
    cleaned_amount_str = meal_amount_raw.replace("NT$", "").replace(",", "").strip()
    meal_amount = float(cleaned_amount_str)

    # 將人數文字轉成整數
    people = int(people_raw)

    # 3. 分步計算費用
    service_fee = meal_amount * service_rate
    total_amount = meal_amount + service_fee - discount
    cost_per_person = total_amount / people

    # 4. 判斷是否超過預算
    is_over_budget = cost_per_person > budget_per_person

    # 5. 輸出費用摘要
    print(f"聚餐名稱：{event_name}")
    print(f"原始餐費：{meal_amount:.2f} 元")
    print(f"服務費：{service_fee:.2f} 元")
    print(f"折價券：{discount:.2f} 元")
    print(f"應付總額：{total_amount:.2f} 元")
    print(f"分攤人數：{people} 人")
    print(f"每人費用：{cost_per_person:.2f} 元")
    print(f"每人預算：{budget_per_person:.2f} 元")
    print(f"是否超過預算：{is_over_budget}")
    print("-" * 30)

# ------------------------------------------
# 測試執行
# ------------------------------------------

# 共用參數
service_rate = 0.10
budget_per_person = 300

print("【測試一】使用原始資料：4 人、折價券 100 元")
calculate_dinner_costs(
    event_name_raw="  WEEKEND DINNER  ",
    meal_amount_raw=" NT$1,200 ",
    people_raw="4",
    service_rate=service_rate,
    discount=100,
    budget_per_person=budget_per_person
)

print("\n【測試二】只將人數改成 '5'，折價券維持 100 元")
calculate_dinner_costs(
    event_name_raw="  WEEKEND DINNER  ",
    meal_amount_raw=" NT$1,200 ",
    people_raw="5",
    service_rate=service_rate,
    discount=100,
    budget_per_person=budget_per_person
)

print("\n【測試三】人數恢復為 '4'，只將折價券改成 220 元")
calculate_dinner_costs(
    event_name_raw="  WEEKEND DINNER  ",
    meal_amount_raw=" NT$1,200 ",
    people_raw="4",
    service_rate=service_rate,
    discount=220,
    budget_per_person=budget_per_person
)
```
